<div class="code-rationale">
Now we define helper utilities. These functions abstract shell command execution for cross-platform compatibility. Not everyone has gcloud installed or uses the same OS.
<br><br>
<strong>Important functions:</strong><br/>
• <code>resolve_bin</code> searches for gcloud across different OS naming conventions<br>
• <code>run_cmd</code> provides consistent error handling and output capture<br>
• Falls back to Python APIs when gcloud is unavailable
</div>

```python
# Helper utilities for shell commands and file operations
import shutil, subprocess, json, sys
from pathlib import Path
from typing import Optional

def which(cmd: str): 
    return shutil.which(cmd)

def resolve_bin(candidates):
    """Find executable across different OS naming conventions"""
    for name in candidates:
        p = which(name)
        if p: return p
    return None

def run_cmd(cmd, check=True):
    """Execute shell command with proper error handling"""
    print("$", " ".join(cmd))
    p = subprocess.run(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
    print(p.stdout)
    if check and p.returncode != 0:
        raise RuntimeError(f"Command failed: {' '.join(cmd)}")
    return p

GCLOUD = resolve_bin(["gcloud", "gcloud.cmd", "gcloud.exe", "gcloud.bat"])

def run_gcloud(args, check=True):
    """Wrapper for gcloud commands"""
    if not GCLOUD: 
        raise FileNotFoundError("gcloud not found on PATH")
    return run_cmd([GCLOUD] + args, check=check)

print(f"✅ gcloud CLI found at: {GCLOUD}" if GCLOUD else "⚠️ gcloud not found (will use Python APIs)")
```

<div class="code-rationale">
We need cloud storage for assets and BigQuery for processing. Creating them programmatically ensures consistent configuration.
<br><br>
<strong>Things to note:</strong><br/>
• <code>uniform_bucket_level_access</code> simplifies IAM and improves security<br>
• <code>exists_ok=True</code> makes the script idempotent (safe to run multiple times)<br>
• Try/except blocks handle resources that already exist gracefully
</div>

```python
# Create GCS bucket and BigQuery dataset
from google.cloud import storage, bigquery
from google.api_core.exceptions import Conflict, NotFound

print("☁️ Creating cloud resources...")
print("=" * 40)

# Create storage bucket
storage_client = storage.Client(project=PROJECT_ID)
try:
    storage_client.get_bucket(BUCKET_NAME)
    print(f"✓ Bucket exists: {BUCKET_NAME}")
except Exception:
    bucket = storage.Bucket(storage_client, name=BUCKET_NAME)
    bucket.iam_configuration.uniform_bucket_level_access_enabled = True
    storage_client.create_bucket(bucket, location="US")
    print(f"✅ Created bucket: {BUCKET_NAME}")

# Create BigQuery dataset
bq = bigquery.Client(project=PROJECT_ID, location=BQ_LOCATION)
ds_ref = bigquery.Dataset(f"{PROJECT_ID}.{DATASET_ID}")
try:
    bq.get_dataset(ds_ref)
    print(f"✓ Dataset exists: {DATASET_ID}")
except NotFound:
    bq.create_dataset(ds_ref, exists_ok=True)
    print(f"✅ Created dataset: {DATASET_ID}")

print("\n✅ Cloud resources ready!")
```

<div class="code-rationale">
BigQuery needs a secure way to access Vertex AI models and Cloud Storage. The connection provides a service account with appropriate permissions.
<br><br>
<strong>Things to note:</strong><br/>
• Service account is automatically created and managed by Google<br>
• Minimal IAM permissions (Object Viewer for storage, AI User for Vertex)<br>
• Policy updates are idempotent to avoid duplicate role bindings
</div>

```python
# Create Cloud Resource Connection for AI model access
from google.cloud import bigquery_connection_v1, storage as _storage, resourcemanager_v3

print("🔗 Setting up AI model connection...")
print("=" * 40)

conn_client = bigquery_connection_v1.ConnectionServiceClient()
parent = f"projects/{PROJECT_ID}/locations/{BQ_LOCATION}"
conn_name = f"{parent}/connections/{CONNECTION_ID}"

# Get or create connection
try:
    conn = conn_client.get_connection(name=conn_name)
    print(f"✓ Connection exists: {CONNECTION_ID}")
except Exception:
    req = bigquery_connection_v1.CreateConnectionRequest(
        parent=parent,
        connection_id=CONNECTION_ID,
        connection=bigquery_connection_v1.Connection(
            cloud_resource=bigquery_connection_v1.CloudResourceProperties()
        )
    )
    conn = conn_client.create_connection(request=req)
    print(f"✅ Created connection: {CONNECTION_ID}")

# Get service account
sa = conn_client.get_connection(name=conn_name).cloud_resource.service_account_id
print(f"\n📧 Service account: {sa}")

# Grant necessary permissions
print("\n🔐 Configuring permissions...")

# Grant Storage Object Viewer
storage_client = _storage.Client(project=PROJECT_ID)
bucket = storage_client.bucket(BUCKET_NAME)
policy = bucket.get_iam_policy(requested_policy_version=3)
member = f"serviceAccount:{sa}"
role = "roles/storage.objectViewer"

# Check if role already exists to avoid duplicates
updated = False
for b in policy.bindings:
    if b["role"] == role:
        if member not in b["members"]:
            b["members"].append(member)
            updated = True
        break
else:
    policy.bindings.append({"role": role, "members": [member]})
    updated = True

if updated:
    bucket.set_iam_policy(policy)
    print("  ✅ Granted Storage Object Viewer")
else:
    print("  ✓ Storage permissions already set")

# Grant Vertex AI User
if GCLOUD:
    run_gcloud([
        "projects", "add-iam-policy-binding", PROJECT_ID,
        "--member", member, 
        "--role", "roles/aiplatform.user"
    ], check=False)
    print("  ✅ Granted Vertex AI User")
else:
    print("  ⚠️ Please grant 'Vertex AI User' role in Console")

print("\n✅ Project setup complete!")
```

## <span class="step-indicator">4</span> Asset Discovery & Validation

<div class="code-rationale">
We need to ensure assets exist before attempting to process them. This prevents cryptic errors later in the pipeline. This notebook assumes your assets are available locally in a folder.
<br><br>
<strong>Things to note:</strong><br/>
• <code>glob("*")</code> for flexible file matching<br>
• Size calculations help estimate processing time and costs<br>
• Assertions provide clear error messages if assets are missing
</div>

```python
# Discover and validate local assets
from pathlib import Path

print("📂 Scanning local assets...")
print("=" * 40)

# Scan directories
img_files = list(Path(IMAGES_DIR).glob("*"))
vid_files = list(Path(VIDEOS_DIR).glob("*"))
pdf_files = list(Path(MANUALS_DIR).glob("*.pdf"))

# Calculate sizes
def get_size_mb(files):
    return sum(f.stat().st_size for f in files if f.is_file()) / (1024 * 1024)

total_files = len(pdf_files) + len(img_files) + len(vid_files)
total_size = get_size_mb(pdf_files) + get_size_mb(img_files) + get_size_mb(vid_files)

# Display inventory
print(f"\n📊 Asset Inventory:")
print(f"  • {len(pdf_files)} PDF manuals ({get_size_mb(pdf_files):.1f} MB)")
print(f"  • {len(img_files)} images ({get_size_mb(img_files):.1f} MB)")
print(f"  • {len(vid_files)} videos ({get_size_mb(vid_files):.1f} MB)")
print(f"  • Total: {total_files} files ({total_size:.1f} MB)")

# Validation with helpful error messages
assert pdf_files, f"⚠️ No PDF manuals found! Please add at least one PDF to: {MANUALS_DIR}"
assert img_files or vid_files, f"⚠️ No images or videos found! Please add files to: {IMAGES_DIR} or {VIDEOS_DIR}"

print("\n✅ Assets validated and ready for processing!")
```

## <span class="step-indicator">5</span> Cloud Upload & Manifest Creation

<div class="code-rationale">
We create a CSV manifest that tracks the mapping between local files and cloud URIs, essential for debugging and audit trails. Then we upload the assets to the bucket.
<br><br>
<strong>Things to note:</strong><br/>
• <code>as_posix()</code> ensures forward slashes even on Windows<br>
• Organized folder structure (images/, videos/, manuals/) in GCS<br>
• Progress tracking for large uploads
</div>

```python
# Upload assets to GCS with manifest creation
import csv
from google.cloud import storage
from pathlib import Path

print("☁️ Uploading assets to cloud storage...")
print("=" * 40)

# Build manifest for tracking
manifest_path = Path(LOCAL_DATA_DIR)/"assets_manifest.csv"
with open(manifest_path, "w", newline="") as f:
    w = csv.writer(f)
    w.writerow(["modality", "local_path", "gcs_uri"])
    
    # Collect all files for upload
    upload_queue = []
    
    for p in Path(IMAGES_DIR).glob("*"):
        if p.is_file():
            w.writerow(["image", str(p), f"{BUCKET_URI}/images/{p.name}"])
            upload_queue.append((p, f"images/{p.name}"))
    
    for p in Path(MANUALS_DIR).glob("*.pdf"):
        w.writerow(["manual_pdf", str(p), f"{BUCKET_URI}/manuals/{p.name}"])
        upload_queue.append((p, f"manuals/{p.name}"))
    
    for p in Path(VIDEOS_DIR).glob("*"):
        if p.is_file():
            w.writerow(["video", str(p), f"{BUCKET_URI}/videos/{p.name}"])
            upload_queue.append((p, f"videos/{p.name}"))

print(f"📋 Created manifest: {manifest_path}")
print(f"📤 Uploading {len(upload_queue)} files...")

# Upload with progress tracking
client = storage.Client(project=PROJECT_ID)
bucket = client.bucket(BUCKET_NAME)

for i, (local_path, cloud_path) in enumerate(upload_queue, 1):
    # Upload file (use forward slashes for GCS paths)
    bucket.blob(cloud_path).upload_from_filename(str(local_path))
    
    # Show progress
    if i % 5 == 0 or i == len(upload_queue):
        print(f"  Progress: {i}/{len(upload_queue)} files uploaded")

print(f"\n✅ Successfully uploaded {len(upload_queue)} files to {BUCKET_URI}")
```

## <span class="step-indicator">6</span> AI-Powered Indexing

<div class="code-rationale">
BigQuery Object Tables provide a SQL interface to cloud storage files, enabling batch processing without data movement.
<br><br>
<strong>Things to note:</strong><br/>
• <code>EXTERNAL TABLE</code> keeps data in GCS (cost-effective)<br>
• <code>object_metadata = 'SIMPLE'</code> includes file metadata<br>
• Wildcard patterns for automatic inclusion of new files
</div>

```python
# Register BigQuery Object Tables for cloud assets
from google.cloud import bigquery

print("📊 Creating BigQuery object tables...")
print("=" * 40)

bq = bigquery.Client(project=PROJECT_ID, location=BQ_LOCATION)

tables = [
    ("images", "images/*", "Image reference library"),
    ("videos", "videos/*", "Training video library"),
    ("manual_pdfs", "manuals/*", "PDF manual library")
]

for table_name, uri_pattern, description in tables:
    sql = f"""
    CREATE OR REPLACE EXTERNAL TABLE `{PROJECT_ID}.{DATASET_ID}.{table_name}`
    WITH CONNECTION `{PROJECT_ID}.{BQ_LOCATION}.{CONNECTION_ID}`
    OPTIONS (
        object_metadata = 'SIMPLE',
        uris = ['{BUCKET_URI}/{uri_pattern}']
    );
    """
    bq.query(sql).result()
    print(f"  ✅ {table_name}: {description}")

print("\n✅ Object tables created successfully!")
```

• Error handling for problematic PDF pages
</div>

```python
# Parse and chunk PDF manuals for semantic search
from pypdf import PdfReader
import pandas as pd
from google.cloud import bigquery
from pathlib import Path

print("📚 Processing PDF manuals...")
print("=" * 40)

bq = bigquery.Client(project=PROJECT_ID, location=BQ_LOCATION)
pdf_files = sorted(Path(MANUALS_DIR).glob("*.pdf"))

if pdf_files:
    pdf_path = pdf_files[0]
    print(f"\nParsing: {pdf_path.name}")
    
    # Extract text from PDF
    reader = PdfReader(str(pdf_path))
    print(f"  Pages: {len(reader.pages)}")
    
    # Extract text from each page
    paras = []
    for i, page in enumerate(reader.pages):
        try:
            text = page.extract_text() or ""
            paras.append(text)
        except Exception as e:
            print(f"  ⚠️ Error on page {i}: {e}")
            paras.append("")
    
    text = "\n\n".join(paras)
    print(f"  Characters extracted: {len(text):,}")
    
    # Smart chunking with overlap
    chunks = []
    chunk_size = 1000
    
    paragraphs = [p.strip() for p in text.split("\n\n") if p.strip()]
    
    current_chunk = []
    current_size = 0
    
    for para in paragraphs:
        para_size = len(para)
        
        if current_size + para_size > chunk_size and current_chunk:
            # Save current chunk
            chunks.append(" ".join(current_chunk))
            
            # Start new chunk with overlap (keep last paragraph)
            if len(current_chunk) > 1:
                current_chunk = current_chunk[-1:]
                current_size = len(current_chunk[0])
            else:
                current_chunk = []
                current_size = 0
        
        current_chunk.append(para)
        current_size += para_size
    
    if current_chunk:
        chunks.append(" ".join(current_chunk))
    
    print(f"  Chunks created: {len(chunks)}")
    print(f"  Avg chunk size: {sum(len(c) for c in chunks) / len(chunks):.0f} chars")
    
    # Create DataFrame
    df = pd.DataFrame({
        "asset_id": [pdf_path.stem] * len(chunks),
        "section": [f"chunk-{i:04d}" for i in range(len(chunks))],
        "chunk_text": chunks,
        "source_uri": [f"{BUCKET_URI}/manuals/{pdf_path.name}"] * len(chunks)
    })
    
    # Load to BigQuery
    table_id = f"{PROJECT_ID}.{DATASET_ID}.manual_chunks"
    job = bq.load_table_from_dataframe(df, table_id)
    job.result()
    
    print(f"\n✅ Loaded {len(df)} text chunks to BigQuery")
    
    # Show sample
    print("\n📝 Sample chunk:")
    print("─" * 40)
    print(chunks[0][:300] + "..." if chunks else "No chunks")
else:
    print("⚠️ No PDF files found to process")
```

<div class="code-rationale">
Embeddings convert content into numerical vectors that capture semantic meaning, enabling similarity search across different modalities.
<br><br>
<strong>Things to note:</strong><br/>
• <code>flatten_json_output</code> for easier processing<br>
• Explicit 1408 dimensions for consistency<br>
• Video processing with 15-second windows and 5-second stride<br>
• Filtering for valid content types and non-empty text
</div>

```python
# Generate AI embeddings for all modalities
from google.cloud import bigquery
import time

print("🧠 Generating AI embeddings...")
print("=" * 40)
print("This may take a few minutes for large datasets\n")

bq = bigquery.Client(project=PROJECT_ID, location=BQ_LOCATION)

# Image embeddings
print("⏳ Processing image embeddings...")
image_sql = f"""
CREATE OR REPLACE TABLE `{PROJECT_ID}.{DATASET_ID}.image_embeddings` AS
SELECT
    uri AS gcs_uri,
    REGEXP_EXTRACT(uri, r'[^/]+$') AS asset_id,
    content_type,
    ml_generate_embedding_result AS embedding
FROM ML.GENERATE_EMBEDDING(
    MODEL `{PROJECT_ID}.{DATASET_ID}.mm_embedding`,
    (SELECT * FROM `{PROJECT_ID}.{DATASET_ID}.images` WHERE content_type LIKE 'image/%'),
    STRUCT(TRUE AS flatten_json_output, 1408 AS output_dimensionality)
);
"""
bq.query(image_sql).result()
print("  ✅ Image embeddings complete")

# Text embeddings
print("⏳ Processing text embeddings...")
text_sql = f"""
CREATE OR REPLACE TABLE `{PROJECT_ID}.{DATASET_ID}.text_embeddings` AS
SELECT
    asset_id,
    section,
    content AS chunk_text,
    source_uri AS gcs_uri,
    ml_generate_embedding_result AS embedding
FROM ML.GENERATE_EMBEDDING(
    MODEL `{PROJECT_ID}.{DATASET_ID}.mm_embedding`,
    (
        SELECT
            chunk_text AS content,
            asset_id,
            section,
            source_uri
        FROM `{PROJECT_ID}.{DATASET_ID}.manual_chunks`
        WHERE LENGTH(chunk_text) > 0
    ),
    STRUCT(TRUE AS flatten_json_output, 1408 AS output_dimensionality)
);
"""
bq.query(text_sql).result()
print("  ✅ Text embeddings complete")

# Video embeddings (with temporal segments)
print("⏳ Processing video embeddings...")
video_sql = f"""
CREATE OR REPLACE TABLE `{PROJECT_ID}.{DATASET_ID}.video_embeddings` AS
SELECT
    uri AS gcs_uri,
    REGEXP_EXTRACT(uri, r'[^/]+$') AS asset_id,
    ml_generate_embedding_start_sec AS start_second,
    ml_generate_embedding_end_sec AS end_second,
    ml_generate_embedding_result AS embedding
FROM ML.GENERATE_EMBEDDING(
    MODEL `{PROJECT_ID}.{DATASET_ID}.mm_embedding`,
    TABLE `{PROJECT_ID}.{DATASET_ID}.videos`,
    STRUCT(
        TRUE AS flatten_json_output,
        1408 AS output_dimensionality,
        0 AS start_second,
        15 AS end_second,
        5 AS interval_seconds
    )
);
"""
bq.query(video_sql).result()
print("  ✅ Video embeddings complete")

print("\n✅ All embeddings generated successfully!")
```

<div class="code-rationale">
Combining all embeddings into a single table enables cross-modal search—finding text that matches an image query, for example.
<br><br>
<strong>Things to note:</strong><br/>
• Standardized schema across modalities<br>
• Metadata fields for filtering (product, language)<br>
• Snippet field for quick result preview<br>

```python
# Build unified knowledge corpus
from google.cloud import bigquery

print("🔍 Building unified knowledge base...")
print("=" * 40)

bq = bigquery.Client(project=PROJECT_ID, location=BQ_LOCATION)

# Create unified corpus combining all modalities
corpus_sql = f"""
CREATE OR REPLACE TABLE `{PROJECT_ID}.{DATASET_ID}.kb_corpus` AS
-- Image embeddings
SELECT
    asset_id,
    'image' AS modality,
    'general-tools' AS product,
    NULL AS part_no,
    'en' AS language,
    REGEXP_EXTRACT(gcs_uri, r'[^/]+$') AS snippet,
    gcs_uri,
    embedding
FROM `{PROJECT_ID}.{DATASET_ID}.image_embeddings`
WHERE embedding IS NOT NULL AND ARRAY_LENGTH(embedding) = 1408

UNION ALL
-- Text embeddings from manuals
SELECT
    asset_id,
    'text',
    'general-tools',
    NULL,
    'en',
    SUBSTR(chunk_text, 1, 300) AS snippet,
    gcs_uri,
    embedding
FROM `{PROJECT_ID}.{DATASET_ID}.text_embeddings`
WHERE embedding IS NOT NULL AND ARRAY_LENGTH(embedding) = 1408

UNION ALL
-- Video segment embeddings
SELECT
    CONCAT(asset_id, '#', CAST(start_second AS STRING), '-', CAST(end_second AS STRING)) AS asset_id,
    'video_segment',
    'general-tools',
    NULL,
    'en',
    CONCAT('segment ', CAST(start_second AS STRING), '-', CAST(end_second AS STRING)) AS snippet,
    gcs_uri,
    embedding
FROM `{PROJECT_ID}.{DATASET_ID}.video_embeddings`
WHERE embedding IS NOT NULL AND ARRAY_LENGTH(embedding) = 1408
;
"""

bq.query(corpus_sql).result()
print("✅ Knowledge corpus created")

# Verify corpus statistics
stats_sql = f"""
SELECT
    modality,
    COUNT(*) AS items,
    COUNTIF(embedding IS NULL) AS null_embeddings,
    MIN(ARRAY_LENGTH(embedding)) AS min_dim,
    MAX(ARRAY_LENGTH(embedding)) AS max_dim
FROM `{PROJECT_ID}.{DATASET_ID}.kb_corpus`
GROUP BY modality
ORDER BY modality;
"""

stats_df = bq.query(stats_sql).result().to_dataframe()
print("\n📊 Corpus Statistics:")
print(stats_df.to_string(index=False))

# Determine indexing strategy based on corpus size
total_rows_sql = f"SELECT COUNT(*) AS c FROM `{PROJECT_ID}.{DATASET_ID}.kb_corpus`"
total_rows = list(bq.query(total_rows_sql).result())[0]['c']

print(f"\n📈 Total corpus size: {total_rows} items")

# Create appropriate index for performance
if total_rows >= 5000:
    print("\n🎯 Creating IVF index for fast similarity search...")
    index_sql = f"""
    CREATE OR REPLACE VECTOR INDEX kb_ivf
    ON `{PROJECT_ID}.{DATASET_ID}.kb_corpus`(embedding)
    STORING (product, language, modality, snippet, gcs_uri)
    OPTIONS(index_type='IVF', distance_type='COSINE');
    """
    bq.query(index_sql).result()
    ANN_OPTIONS = '{"fraction_lists_to_search": 0.01}'
    print("  ✅ IVF index created for optimized search")
else:
    ANN_OPTIONS = '{"use_brute_force": true}'
    print("  ℹ️ Using brute force search (corpus < 5000 items)")

# Create text search index for snippet field
search_index_sql = f"""
CREATE SEARCH INDEX IF NOT EXISTS idx_snippet
ON `{PROJECT_ID}.{DATASET_ID}.kb_corpus`(snippet);
"""
bq.query(search_index_sql).result()

print("\n✅ Knowledge base ready for queries!")
```

<div class="code-rationale">
Here we define a visualization with local file preview. This assumes you are running locally.
<br><br>
<strong>Things to note:</strong><br/>
• Local file matching enables image/video preview<br>
• Video segments show with start time for quick navigation<br>
• Similarity scores converted to percentages for intuition<br>
• Modality badges for quick content type identification<br>
• HTML5 video controls with JavaScript seek functionality

```python
# Enhanced visualization helper for multimodal results with local file display
from IPython.display import display, HTML, Image
from pathlib import Path
import pandas as pd
import base64, mimetypes, uuid, re

BASE = Path(LOCAL_DATA_DIR)

def _find_local(filename: str):
    """Find local file in data directories"""
    for sub in ("images", "videos", "manuals"):
        p = BASE / sub
        if p.exists():
            for f in p.rglob(filename):
                if f.is_file():
                    return f
    return None

def _basename(uri: str) -> str:
    """Extract filename from URI"""
    try:
        return Path(uri).name
    except:
        return str(uri).split("/")[-1]

def _extract_start_sec(rec: pd.Series) -> int:
    """Extract video start second from record"""
    # Try explicit fields first
    for key in ("start_second", "start", "offset", "start_sec", "segment_start"):
        if key in rec.index:
            try:
                v = rec.get(key)
                if pd.notna(v):
                    return int(float(v))
            except:
                pass
    
    # Try parsing from snippet like "segment 5-15"
    snip = str(rec.get("snippet", ""))
    m = re.search(r'(\d+(?:\.\d+)?)\s*-\s*(\d+(?:\.\d+)?)', snip)
    if m:
        try:
            return int(float(m.group(1)))
        except:
            pass
    
    # Try parsing from asset_id like "file#5-15"
    aid = str(rec.get("asset_id", ""))
    m = re.search(r'#\s*(\d+(?:\.\d+)?)\s*-\s*(\d+(?:\.\d+)?)', aid)
    if m:
        try:
            return int(float(m.group(1)))
        except:
            pass
    return 0

def _html_video_with_start(local_path: Path, start_second: int, width: int = 240) -> str:
    """Create HTML5 video element with automatic seek to start time"""
    mime, _ = mimetypes.guess_type(str(local_path))
    if not mime:
        mime = "video/mp4"
    
    # Create base64 data URI for embedded video
    b64 = base64.b64encode(local_path.read_bytes()).decode("ascii")
    src = f"data:{mime};base64,{b64}"
    vid = f"vid_{uuid.uuid4().hex[:8]}"
    
    html = f"""
    <div style='margin: 1rem 0'>
      <video id="{vid}" controls preload="metadata" style="max-width:{width}px; width:100%; border:1px solid #ddd; border-radius:6px;">
        <source src="{src}" type="{mime}">
        Your browser does not support HTML5 video.
      </video>
      <div style="margin-top: 0.5rem; font-size: 0.9rem; color: #666;">
        📍 Segment starts at {start_second} seconds
      </div>
    </div>
    <script>
    (function() {{
      var v = document.getElementById("{vid}");
      if (v) {{
        v.addEventListener('loadedmetadata', function() {{
          if ({start_second} > 0) {{
            v.currentTime = {start_second};
          }}
        }});
      }}
    }})();
    </script>
    """
    return html

def viz_any(df: pd.DataFrame, limit: int = 5):
    """Visualize multimodal search results with rich formatting and local file display"""
    
    if df is None or len(df) == 0:
        display(HTML('<div class="callout warning">No results found. Try adjusting your query.</div>'))
        return
    
    # Deduplicate results
    keys = []
    if "gcs_uri" in df.columns:
        keys.append("gcs_uri")
    if "start_second" in df.columns:
        keys.append("start_second")
    if keys:
        df_view = df.drop_duplicates(subset=keys, keep="first").head(limit)
    else:
        df_view = df.head(limit)
    
    # Start results container
    html = '<div style="margin-top: 1rem;">'
    html += '<h3>🎯 Search Results</h3>'
    
    # Process each result
    for i, (_, row) in enumerate(df_view.iterrows(), 1):
        modality = str(row.get("modality", "")).lower()
        uri = str(row.get("gcs_uri", row.get("uri", "")))
        name = _basename(uri)
        snippet = str(row.get("snippet", ""))[:400]
        
        # Calculate similarity score
        distance = row.get("distance", row.get("similarity", 0))
        if isinstance(distance, (int, float)):
            similarity = 1 - distance if "distance" in row.index else distance
            score_pct = max(0, min(100, similarity * 100))
        else:
            score_pct = 0
        
        # Determine badge style
        badge_class = {
            "image": "image",
            "video_segment": "video",
            "text": "text"
        }.get(modality, "text")
        
        # Build result card
        html += f'''
        <div class="result-card">
            <div style="display: flex; justify-content: space-between; align-items: start; margin-bottom: 0.5rem;">
                <div>
                    <span style="font-size: 1.2rem; font-weight: 600;">#{i}</span>
                    <span class="badge {badge_class}">{modality.replace("_", " ").title()}</span>
                </div>
                <div style="text-align: right;">
                    <div style="font-size: 1.5rem; font-weight: bold; color: #34a853;">{score_pct:.0f}%</div>
                    <div style="font-size: 0.75rem; color: #666;">match</div>
                </div>
            </div>
            <div style="margin: 0.5rem 0;">
                <strong>{name}</strong><br>
                <code style="font-size: 0.8rem; color: #666;">{uri}</code>
            </div>
        '''
        
        # Add content preview based on modality
        local = _find_local(name)
        
        if modality == "text":
            html += f'<div style="padding: 0.5rem; background: #f8f9fa; border-radius: 4px; margin-top: 0.5rem;">{snippet}</div>'
        
        elif modality == "image" and local and local.exists():
            # Display local image
            html += '</div>'  # Close card
            display(HTML(html))
            display(Image(filename=str(local), width=200))
            html = ''  # Reset for next card
        
        elif modality == "video_segment" and local and local.exists():
            # Display local video with seek to segment
            start_sec = _extract_start_sec(row)
            video_html = _html_video_with_start(local, start_sec)
            html += video_html
        
        else:
            html += f'<div style="margin-top: 0.5rem;"><em>{modality.title()} content: {snippet[:100]}...</em></div>'
        
        if html:  # Close card if not already closed
            html += '</div>'
    
    if html:
        html += '</div>'
        display(HTML(html))
    
    # Display summary
    summary = f'''
    <div class="callout info" style="margin-top: 1rem;">
        <strong>📊 Results Summary</strong><br>
        Found {len(df)} total matches • Showing top {min(limit, len(df))} results
    </div>
    '''
    display(HTML(summary))

# Set default search parameters
top_k = 50
try:
    ANN_OPTIONS
except NameError:
    ANN_OPTIONS = '{"use_brute_force": true}'

print("✅ Visualization helper loaded")
print(f"   Search parameters: top_k={top_k}, ANN_OPTIONS={ANN_OPTIONS}")
```

```python
from google.cloud import bigquery
bq = bigquery.Client(project=PROJECT_ID, location=BQ_LOCATION)

print("🎥 Using first video (0-15s) as query\n")

# Generate video query embedding (averaging segments)
print("⏳ Analyzing video motion patterns...")
bq.query(f"""
CREATE OR REPLACE TABLE `{PROJECT_ID}.{DATASET_ID}.query_vec_video` AS
WITH segs AS (
    SELECT ml_generate_embedding_result AS emb
    FROM ML.GENERATE_EMBEDDING(
        MODEL `{PROJECT_ID}.{DATASET_ID}.mm_embedding`,
        (SELECT * FROM `{PROJECT_ID}.{DATASET_ID}.videos` ORDER BY uri LIMIT 1),
        STRUCT(TRUE AS flatten_json_output, 0 AS start_second, 15 AS end_second, 5 AS interval_seconds)
    )
)
SELECT
    ARRAY_AGG(avg_component ORDER BY pos) AS embedding
FROM (
    SELECT pos, AVG(val) AS avg_component
    FROM segs, UNNEST(emb) AS val WITH OFFSET pos
    GROUP BY pos
);
""").result()

# Search for similar patterns
print("🔎 Searching for similar patterns...\n")
search_sql = f"""
SELECT
    base.asset_id,
    base.modality,
    base.snippet,
    base.gcs_uri,
    distance
FROM VECTOR_SEARCH(
    (
        SELECT asset_id, modality, snippet, gcs_uri, embedding
        FROM `{PROJECT_ID}.{DATASET_ID}.kb_corpus`
        WHERE product = 'general-tools' AND language = 'en'
    ),
    'embedding',
    (SELECT embedding FROM `{PROJECT_ID}.{DATASET_ID}.query_vec_video`),
    top_k => {top_k},
    distance_type => 'COSINE'
)
ORDER BY distance ASC;
"""

results_df = bq.query(search_sql).result().to_dataframe()
viz_any(results_df)
```

## Scenario 4: Text + Image Combo → Any

```python
from google.cloud import bigquery
bq = bigquery.Client(project=PROJECT_ID, location=BQ_LOCATION)

print("🔀 Combining text description + photo evidence\n")

# Search using both text and image, merge results
combo_sql = f"""
WITH
text_hits AS (
    SELECT
        base.asset_id, base.modality, base.snippet, base.gcs_uri, 
        distance, 'text' AS query_source
    FROM VECTOR_SEARCH(
        (
            SELECT asset_id, modality, snippet, gcs_uri, embedding
            FROM `{PROJECT_ID}.{DATASET_ID}.kb_corpus`
            WHERE product = 'general-tools' AND language = 'en'
        ),
        'embedding',
        (SELECT embedding FROM `{PROJECT_ID}.{DATASET_ID}.query_vec_text`),
        top_k => 100,
        distance_type => 'COSINE'
    )
),
image_hits AS (
    SELECT
        base.asset_id, base.modality, base.snippet, base.gcs_uri, 
        distance, 'image' AS query_source
    FROM VECTOR_SEARCH(
        (
            SELECT asset_id, modality, snippet, gcs_uri, embedding
            FROM `{PROJECT_ID}.{DATASET_ID}.kb_corpus`
            WHERE product = 'general-tools' AND language = 'en'
        ),
        'embedding',
        (SELECT embedding FROM `{PROJECT_ID}.{DATASET_ID}.query_vec_image`),
        top_k => 100,
        distance_type => 'COSINE'
    )
)
SELECT
    asset_id, modality, snippet, gcs_uri, 
    MIN(distance) AS distance,
    STRING_AGG(DISTINCT query_source, ', ') AS matched_by
FROM (
    SELECT * FROM text_hits
    UNION ALL
    SELECT * FROM image_hits
)
GROUP BY asset_id, modality, snippet, gcs_uri
ORDER BY distance ASC
LIMIT 50;
"""

print("⏳ Running multimodal fusion search...")
results_df = bq.query(combo_sql).result().to_dataframe()

print("\n💡 Results show best matches from BOTH text and image evidence:\n")
viz_any(results_df)
```

```python
from google.cloud import bigquery
bq = bigquery.Client(project=PROJECT_ID, location=BQ_LOCATION)

print("🔀 Combining text description + video evidence\n")

sql_text_video = f"""
WITH
text_hits AS (
    SELECT base.asset_id, base.modality, base.snippet, base.gcs_uri, distance
    FROM VECTOR_SEARCH(
        (SELECT * FROM `{PROJECT_ID}.{DATASET_ID}.kb_corpus`),
        'embedding',
        TABLE `{PROJECT_ID}.{DATASET_ID}.query_vec_text`,
        top_k => {top_k},
        distance_type => 'COSINE',
        options => '{ANN_OPTIONS}'
    )
),
video_hits AS (
    SELECT base.asset_id, base.modality, base.snippet, base.gcs_uri, distance
    FROM VECTOR_SEARCH(
        (SELECT * FROM `{PROJECT_ID}.{DATASET_ID}.kb_corpus`),
        'embedding',
        TABLE `{PROJECT_ID}.{DATASET_ID}.query_vec_video`,
        top_k => {top_k},
        distance_type => 'COSINE',
        options => '{ANN_OPTIONS}'
    )
),
merged AS (
    SELECT asset_id,
           ANY_VALUE(modality) AS modality,
           ANY_VALUE(snippet) AS snippet,
           ANY_VALUE(gcs_uri) AS gcs_uri,
           MIN(distance) AS distance
    FROM (
        SELECT * FROM text_hits
        UNION ALL
        SELECT * FROM video_hits
    )
    GROUP BY asset_id
)
SELECT * FROM merged
ORDER BY distance ASC
LIMIT {top_k}
"""

print("⏳ Searching with text + video evidence...")
results_df = bq.query(sql_text_video).result().to_dataframe()
viz_any(results_df)
```

## Scenario 7: Text + Image + Video → Any

<div class="callout info">
  <strong>🎯 The Situation</strong><br>
  Complex diagnosis requiring all available evidence: symptoms description, component photo, and malfunction video. The system triangulates across all three inputs.
</div>

```python
from google.cloud import bigquery
bq = bigquery.Client(project=PROJECT_ID, location=BQ_LOCATION)

print("🔀 Combining ALL evidence types: text + image + video\n")

sql_all_modalities = f"""
WITH
text_hits AS (
    SELECT base.asset_id, base.modality, base.snippet, base.gcs_uri, distance
    FROM VECTOR_SEARCH(
        (SELECT * FROM `{PROJECT_ID}.{DATASET_ID}.kb_corpus`),
        'embedding',
        TABLE `{PROJECT_ID}.{DATASET_ID}.query_vec_text`,
        top_k => {top_k},
        distance_type => 'COSINE',
        options => '{ANN_OPTIONS}'
    )
),
image_hits AS (
    SELECT base.asset_id, base.modality, base.snippet, base.gcs_uri, distance
    FROM VECTOR_SEARCH(
        (SELECT * FROM `{PROJECT_ID}.{DATASET_ID}.kb_corpus`),
        'embedding',
        TABLE `{PROJECT_ID}.{DATASET_ID}.query_vec_image`,
        top_k => {top_k},
        distance_type => 'COSINE',
        options => '{ANN_OPTIONS}'
    )
),
video_hits AS (
    SELECT base.asset_id, base.modality, base.snippet, base.gcs_uri, distance
    FROM VECTOR_SEARCH(
        (SELECT * FROM `{PROJECT_ID}.{DATASET_ID}.kb_corpus`),
        'embedding',
        TABLE `{PROJECT_ID}.{DATASET_ID}.query_vec_video`,
        top_k => {top_k},
        distance_type => 'COSINE',
        options => '{ANN_OPTIONS}'
    )
),
merged AS (
    SELECT asset_id,
           ANY_VALUE(modality) AS modality,
           ANY_VALUE(snippet) AS snippet,
           ANY_VALUE(gcs_uri) AS gcs_uri,
           MIN(distance) AS distance,
           COUNT(DISTINCT CASE 
               WHEN distance < 0.5 THEN 'strong'
               WHEN distance < 0.7 THEN 'moderate'
               ELSE 'weak'
           END) as evidence_strength
    FROM (
        SELECT * FROM text_hits
        UNION ALL
        SELECT * FROM image_hits
        UNION ALL
        SELECT * FROM video_hits
    )
    GROUP BY asset_id
)
SELECT * FROM merged
ORDER BY distance ASC
LIMIT {top_k}
"""

print("⏳ Running maximum evidence search...")
results_df = bq.query(sql_all_modalities).result().to_dataframe()
print("\n🎯 Results ranked by combined evidence from all sources:\n")
viz_any(results_df)
```
